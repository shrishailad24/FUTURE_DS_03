```python
import pandas as pd

# Load dataset
df = pd.read_csv("../data/Bank_Marketing_Dataset.csv")   # Change the filename if needed

# First 5 rows
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)

print("Duplicate Rows after cleaning:", df.duplicated().sum())
```

```python
print(df.isnull().sum())
```

```python
df.fillna(method="ffill", inplace=True)
```

```python
df.describe(include="all")
```

```python
df.to_csv("../data/cleaned_marketing_data.csv", index=False)

print("✅ Cleaned dataset saved successfully!")
```

```python
import matplotlib.pyplot as plt

df['deposit'].value_counts().plot(kind='bar')
plt.title("Deposit Subscription")
plt.xlabel("Subscribed")
plt.ylabel("Count")
plt.show()
```

```python
df['job'].value_counts().plot(kind='bar', figsize=(10,5))
plt.title("Customers by Job")
plt.xticks(rotation=45)
plt.show()
```

```python
df['age'].plot(kind='hist', bins=20)
plt.title("Age Distribution")
plt.show()
```

```python
df['balance'].plot(kind='hist', bins=30)
plt.title("Account Balance Distribution")
plt.show()
```

```python
df['contact'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Contact Method")
plt.ylabel("")
plt.show()
```

```python
df['month'].value_counts().plot(kind='bar')
plt.title("Campaigns by Month")
plt.xticks(rotation=45)
plt.show()
```
